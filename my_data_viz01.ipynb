## Data visualization

### Subtask:
Create a choropleth map visualizing the 'Happiness score' for each country.


**Reasoning**:
Generate the choropleth map using Plotly, handling potential country name discrepancies.



```python
import plotly.express as px

try:
    fig = px.choropleth(df, locations='Country', locationmode='country names',
                        color='Happiness score', hover_name='Country',
                        title='Global Happiness Score 2022',
                        color_continuous_scale=px.colors.sequential.Plasma)
    fig.show()
except ValueError as e:
    print(f"Error creating choropleth: {e}")
    # Attempt to fix country names
    import re

    def clean_country_name(name):
        name = re.sub(r'\*', '', name)  # remove asterisk
        return name

    df['Country'] = df['Country'].apply(clean_country_name)

    try:
        fig = px.choropleth(df, locations='Country', locationmode='country names',
                            color='Happiness score', hover_name='Country',
                            title='Global Happiness Score 2022',
                            color_continuous_scale=px.colors.sequential.Plasma)
        fig.show()
    except ValueError as e:
        print(f"Error after cleaning country names: {e}")
        print("Failed to create a valid choropleth map.")
```
